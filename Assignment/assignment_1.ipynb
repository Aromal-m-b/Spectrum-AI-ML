# Numpy Practice Questions

```python
import numpy as np
```

### 1. Create a NumPy array containing numbers from 1 to 10.

```python
arr = np.arange(1,11)
print(arr)
print(type(arr))
```

### 2. Create an array of 10 zeros using NumPy.

```python
arr = np.zeros(10,dtype = int)
print(arr)
print(type(arr))
```

### 3. Create an array of 10 ones using NumPy

```python
arr = np.ones(10,dtype = int)
print(arr)
print(type(arr))
```

### 4.  Create an array containing even numbers from 2 to 20 using arange().

```python
arr = np.arange(2,21,2)
print(arr)
print(type(arr))
```

### 5.  Find the dimension, shape, size, and data type of a NumPy array.

```python
arr = np.array([223,5563,4,89,345])
print(arr)
print("Dimension : ",np.ndim(arr))
print("Shape : ",np.shape(arr))
print("Data Type or Array : ",arr.dtype)
```

### 6. Create a 1D array [10, 20, 30, 40, 50] and access the third element.

```python
arr = np.array([10, 20, 30, 40, 50])
print(arr[2])
```

### 7. Create an array [10, 20, 30, 40, 50] and print the elements from index 1 to 3.

```python
arr = np.array([10, 20, 30, 40, 50])
print(arr[1:4])
```

### 8. Create two arrays and perform addition, subtraction, multiplication, and division between them.

```python
arr1 = np.array([1,2,3,4])
arr2 = np.array([5,6,7,8])

print("Addition : ",arr1+arr2)
print("Multiplication : ",arr1*arr2)
print("Subtraction : ",arr1-arr2)
print("Division : ",np.array((arr2/arr1),dtype = 'int64'))
```

### 9. Find the dot product of two NumPy arrays. 

```python
arr1 = np.array([1,2,3,4])
arr2 = np.array([5,6,7,8])

arr = np.dot(arr1,arr2)
print(arr)
```

### 10. Convert a 1D array containing 12 elements into a 3 × 4 matrix using reshape()

```python
arr = np.array([23,45,123,76,123,589,105,12,34,86,5,122])
arr = arr.reshape(3,4)
print(arr)
```

### 11. Flatten a 2D NumPy array into a 1D array

```python
arr = np.array([[23,42,561,944,46],[39,85,10,59,61]])
arr = arr.flatten()
print(arr)
```

### 12.  Create two 2D arrays and combine them using vstack() and hstack()

```python
arr1 = np.array([[23,42,561,944,46],[39,85,10,59,61]])
arr2 = np.array([[12,41,27,89,40],[88,94,75,69,23]])

arr = np.vstack((arr1,arr2))
print("Horizontonal")
print(arr)
print("-----------")
arr = np.hstack((arr1,arr2))
print("Vertical")
print(arr)
print("-----------")
```

### 13.  Given: arr = np.array([10, 25, 30, 45, 60, 75]) — Find all values greater than 40

```python
arr = np.array([10, 25, 30, 45, 60, 75]) 
fit = np.where(arr > 40)
print(arr[fit])
```

### 14. Given: arr = np.array([10, 20, 30, 40, 50]) — Replace all values greater than 30 with 100.

```python
arr = np.array([10, 20, 30, 40, 50])
fit = np.where(arr > 30)
arr[fit] = 100
print(arr)
```

### 15.  Find the second-highest value in a NumPy array.

```python
arr = np.array([23,42,561,944,46])
arr = np.sort(arr)
print("Second Highest Value : ",arr[-2])
```

### 16. Find the second-lowest value in a NumPy array.

```python
arr = np.array([23,42,561,944,46])
arr = np.sort(arr)
print("Second lowest value : ",arr[2])
```

### 17. Remove duplicate values from a NumPy array

```python
arr = np.array([23,42,23,944,561,944,46])
arr = np.unique(arr)
print(arr)
```

### 19.  Find the number of even and odd numbers in an array

```python
arr = np.array([23,42,23,944,561,944,46])
even = np.where(arr%2 == 0)
odd = np.where(arr%2 == 1)
print("No of Even : ",np.size(even))
print("No of Odd : ",np.size(odd))
```
