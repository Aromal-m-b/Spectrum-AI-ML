## 3. Ordinary Z-Score

A **Z-score** tells us how many standard deviations an individual observation is away from the population mean.

Formula:

$$
Z = \frac{X-\mu}{\sigma}
$$

Example:

- `X = 190`
- `μ = 170`
- `σ = 10`

Calculate the Z-score using Python.

````python
X = 190
mu = 170
sigma = 10

Z = (X - mu) / sigma

print("Z-score:", Z)
````

## 4. Real-World Scenario: Customer Spending

An online store wants to understand the average amount spent by its customers.

We do **not** know the true population mean or the true population standard deviation.

We collect a sample of **100 customers**.

The spending values below are manually created for practice. In a real project, these values would normally come from a database, CSV file, API, or another data source.

````python
customer_spending = [
    120, 150, 130, 160, 140, 135, 155, 110, 145, 125,
    170, 132, 148, 138, 152, 128, 142, 158, 136, 144,
    118, 162, 130, 150, 140, 155, 125, 135, 145, 165,
    122, 148, 138, 157, 132, 141, 151, 129, 146, 159,
    134, 143, 153, 127, 139, 161, 131, 147, 156, 124,
    142, 150, 136, 145, 158, 133, 149, 137, 154, 126,
    140, 152, 130, 144, 160, 135, 148, 139, 155, 128,
    143, 151, 132, 146, 157, 121, 138, 149, 134, 153,
    141, 159, 129, 147, 136, 150, 125, 142, 154, 131,
    145, 158, 133, 148, 137, 152, 127, 140, 156, 130
]

print("Number of observations:", len(customer_spending))
````

## 5. Calculate the Sample Mean

The sample mean is:

$$
\bar X = \frac{\sum X_i}{n}
$$

Use Python to calculate the mean of the 100 customers.

````python
n = len(customer_spending)

sample_mean = np.mean(customer_spending)

print("Sample size:", n)
print("Sample mean:", sample_mean)
````

## 6. Calculate the Sample Standard Deviation

Because the population standard deviation `σ` is unknown, we calculate the **sample standard deviation `s`**.

Formula:

$$
s =
\sqrt{
\frac{\sum (X_i-\bar X)^2}{n-1}
}
$$

In NumPy, `ddof=1` tells it to use `n - 1` in the denominator.

````python
sample_std = np.std(customer_spending, ddof=1)

print("Sample standard deviation (s):", sample_std)
````

## 7. Calculate the Standard Error

Since the population standard deviation is unknown, we estimate the standard error using:

$$
SE = \frac{s}{\sqrt n}
$$

Here:

- `s` = sample standard deviation
- `n` = number of observations in our sample

Calculate it using Python.

````python
SE = sample_std / np.sqrt(n)

print("Standard Error:", SE)
````

## 8. Hypothesis Test Scenario

Suppose the business wants to test the claim:

> The average customer spending is **140** currency units.

We define:

$$
H_0: \mu = 140
$$

$$
H_1: \mu \neq 140
$$

We use a significance level of:

$$
\alpha = 0.05
$$

Because the population standard deviation is **unknown**, we will use a **one-sample t-test**.

The seminar notes make this distinction: when `σ` is unknown, the usual inference for a population mean uses the t-distribution rather than the classical Z-test. fileciteturn0file0L454-L475

````python
hypothesized_mean = 140
alpha = 0.05

t_statistic, p_value = stats.ttest_1samp(
    customer_spending,
    popmean=hypothesized_mean
)

print("t-statistic:", t_statistic)
print("p-value:", p_value)
````

## 9. Make the Decision

Decision rule:

$$
p < \alpha \Rightarrow \text{Reject } H_0
$$

$$
p \geq \alpha \Rightarrow \text{Fail to reject } H_0
$$

Run the following cell to automatically make the decision.

````python
if p_value < alpha:
    print("Reject H₀")
    print("There is sufficient evidence that the population mean differs from 140.")
else:
    print("Fail to reject H₀")
    print("There is not sufficient evidence that the population mean differs from 140.")
````

## 10. Connect Everything Together

The complete process is:

```text
100 customer observations
        ↓
Calculate sample mean (X̄)
        ↓
Calculate sample standard deviation (s)
        ↓
Calculate standard error

        SE = s / √n

        ↓
Compare the sample result
with the hypothesized population mean
        ↓
Perform the statistical test
        ↓
Obtain the p-value
        ↓
Compare p with α = 0.05
        ↓
Reject H₀ or fail to reject H₀
```

### Key Idea

We **do not know the population standard deviation `σ`**.

Therefore, from our available sample we calculate:

$$
s = \text{sample standard deviation}
$$

and use:

$$
SE = \frac{s}{\sqrt n}
$$

For the actual hypothesis test, because `σ` is unknown, we use the **t-distribution**.

## 12. Summary

- **Z-score** measures how many standard deviations an individual observation is from a mean.
- **Sample mean `X̄`** is the average of observations in our sample.
- **Sample standard deviation `s`** measures the spread of observations in our sample.
- When population `σ` is unknown, we estimate the standard error using:

$$
SE = \frac{s}{\sqrt n}
$$

- For hypothesis testing about a population mean when `σ` is unknown, the usual method is a **one-sample t-test**.
- The final decision is based on the relationship between the **p-value** and the chosen significance level `α`.
