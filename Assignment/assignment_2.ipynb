```python
def classify_charge(charge, stats):
    best_offer = None
    best_distance = float('inf')

    for offer, values in stats.items():
        mean = values['mean']
        std = values['std']

        z_distance = abs(charge - mean) / std

        if z_distance < best_distance:
            best_distance = z_distance
            best_offer = offer

    return best_offer
```

```python
raw.Offer = raw['Offer'].fillna('Offer Null')
```

```python
offers = list(raw.Offer.unique())
offers.remove('Offer Null')
offers.remove('Offer R')
offers
```

```python
group = raw[raw['Offer'].isin(offers)].groupby('Offer')['Monthly Charge'].agg(['mean','std'])
```

```python
stats = group.to_dict('index')
stats
```

```python
means = {
    offer: values['mean']
    for offer, values in stats.items()
}
means
```

```python
def closest_offer(monthly_charge):
    return min(
        means,
        key=lambda offer: abs(monthly_charge - means[offer])
    )
```

```python
mask = raw['Offer'].isna()

raw.loc[mask, 'Offer'] = raw.loc[mask, 'Monthly Charge'].apply(
    closest_offer
)
```

```python
temp = df.groupby('Age Group')['Churn'].mean()
x = temp.index
y = (temp.values) * 100
sns.barplot(x=x,y=y)
plt.xlabel('Age Group')
plt.ylabel('Churn rate')
plt.title('Churn rate on Age')
plt.show()
```

* **As the Age of the Customer Increases the Customers are choosing to End or Stop company Services**     
* **This behaviour is showing steady increase when it comes to the Elderly**             
* **This spike can also mean that either customer died due to old age or they willingly choose to end Telecome Services**               

```python
fig = px.bar(
    x = x,
    y = y,
    title = 'Churn Rate'
)
fig.show()
```

```python
temp = df.groupby('Married')['Churn'].mean()
fig = px.bar(
    x = temp.index.astype(str),
    y = (temp.values) *  100,
    title = 'Married Churn Rates',
    labels  = 'Churn Rate'
)
fig.update_xaxes(title = 'Married')
fig.update_yaxes(title = 'Churn Rate')
fig.show()
```

```python
df['Tenure Group'] = pd.cut(
    df['Tenure in Months'],
    bins=[0, 6, 12, 24, 48, 72],
    labels=['New', 'Early', 'Established', 'Loyal', 'Long-Term']
)

temp = df.groupby('Tenure Group')['Churn'].mean()
fig = px.bar(
    x = temp.index,
    y = temp.values * 100,
    title = 'Customer Loyalty and Retaining'
)
fig.update_xaxes(title = 'Cutomer Period')
fig.update_yaxes(title  = 'Churn Rates')
fig.show()
```

**The Churn Rate is Decreased overtime, the likelihood of the Loyal Customers Ending their Service is law most of the New customers while Ends thier subscription**
