## Enumeration

```python
import numpy as np
```

### 1D Array

```python
arr = np.array([76,58,90])
for x in arr:
    print(x)
```

### 2D Array

```python
arr = np.array([[34,56,23],[90,7,23]])
for x in arr:
    for y in x:
        print(y)
```

### 3D Array

```python
arr = np.array([[[23,13,56],[7,34,65]],[[4,90,34],[56,32,89]]])
for x in arr:
    for y in x:
        for z in y:
            print(z)
```

### Any number of dimension

```python
arr = np.array([[[23,13,56],[7,34,65]],[[4,90,34],[56,32,89]]])
for x in np.nditer(arr):
    print(x)
```

## Join

### Concatenate

```python
arr1 = np.array([[45,87],[23,75]])
arr2 = np.array([[34,87],[94,74]])
arr = np.concatenate((arr1,arr2),axis = 0)
print(arr)
```

```python
arr1 = np.array([[45,87],[23,75]])
arr2 = np.array([[34,87],[94,74]])
arr = np.concatenate((arr1,arr2),axis = 1)
print(arr)
```

### Stack

```python
arr1 = np.array([64,89,34])
arr2 = np.array([76,86,57])

arr = np.stack((arr1,arr2),axis = 1)
print(arr)
```

```python
arr1 = np.array([64,89,34])
arr2 = np.array([76,86,57])

arr = np.stack((arr1,arr2),axis = 0)
print(arr)
```

### Horizontonal Stack / Row Stack

```python
arr1 = np.array([64,89,34])
arr2 = np.array([76,86,57])

arr = np.hstack((arr1,arr2))
print(arr)
```

### Vertical Stack / Column Stack

```python
arr1 = np.array([64,89,34])
arr2 = np.array([76,86,57])

arr = np.vstack((arr1,arr2))
print(arr)
```

### Depth Stack / Height Stack

```python
arr1 = np.array([64,89,34])
arr2 = np.array([76,86,57])

arr = np.dstack((arr1,arr2))
print(arr)
```

## Splitting

### 1D Split

```python
arr = np.array([98,46,67,12,90,34])
newarr = np.array_split(arr,4)
for x in newarr:
    print(x)
```

### 2D Split

```python
arr = np.array([[23,54],[67,78],[34,98],[12,56],[34,45],[34,21]])
newarr = np.array_split(arr,6)
for x in newarr:
    print(x)
```

### 3D Split

```python
arr = np.array([[[1,2,3]]])
newarr = np.array_split(arr,2)
for x in newarr:
    print(x)
```

### Splitting on Axis

```python
arr = np.array([[23,54],[67,78],[34,98],[12,56],[34,45],[34,21]])
print(arr)
print("\n\n")
newarr = np.array_split(arr,3,axis = 1)
for x in newarr:
    print(x)
    print("\n")
```

### HSplit 

```python
arr = np.array([[34,56,21],[48,8496,2682],[833,849,428],[278,48,40]])
newarr = np.hsplit(arr,3)
for x in newarr:
    print(x)
```

### VSplit

```python
arr = np.array([[34,56,21],[48,8496,2682],[833,849,428],[278,48,40]])
newarr = np.vsplit(arr,2)
for x in newarr:
    print(x)
```

### DSplit

```python
arr = np.array([[[34,56,21],[48,8496,2682]],[[833,849,428],[278,48,40]]])
newarr = np.dsplit(arr,3)
for x in newarr:
    print(x)
    print("------------")
```

## Search

### Where

```python
arr = np.array([2,467,2362,548,708,2637,59,152])
res = np.where(arr == 467)
print(res)
```

```python
res = np.where(arr%2 == 0)
print(res)
```

```python
res = np.where((arr**2)==arr+arr)
print(res)
```

### searchsorted 

```python
arr = np.array([2,3,467,2362,548,708,2637,59,152])
res = np.searchsorted(arr,3)
print(res)
```

```python
res = np.searchsorted(arr,3,side = 'left')
print(res)
```

```python
res = np.searchsorted(arr,3,side = 'right')
print(res)
```

```python
res = np.searchsorted(arr,[4367,1,4])
print(res)
```

## Filter

```python
arr = np.array([743,567,17,74])
x = [True,False,True,True]
print(arr[x])
```

## Ones & Zeros

### 1D Array

```python
arr = np.ones(3)
print(arr)

arr = np.zeros(3)
print(arr)
```

### 2D Array

```python
arr = np.ones((2,3))
print(arr)

arr = np.zeros((2,3))
print(arr)
```

### 3D Array

```python
arr = np.ones((2,3,4),dtype = int)
print(arr)
print("---------")
arr = np.zeros((2,3,4),dtype = int)
print(arr)
print("---------")
```
