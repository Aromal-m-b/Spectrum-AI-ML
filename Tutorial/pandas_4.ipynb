```python
names = df['Name'].tolist()
for i,name in enumerate(names):
    if df.loc[i,['Year']].isnull()['Year']:
        if df['Name'][i][-4:].isdigit():
            df['Year'][i] = float(df['Name'][i][-4:])
        else:
            df['Year'][i] = np.nan
    else:
        df['Year'][i] = np.nan
```

```python
df['Year'] = pd.to_datetime(df['Year'],format='%Y').dt.year.astype('Int64')
df['Year']
```
