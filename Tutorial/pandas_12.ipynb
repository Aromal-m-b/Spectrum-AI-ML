### order_id

```python
df.order_id = df.order_id.str.replace('ORD','')
```

```python
df.order_id = df.order_id.str.replace('-','')
```

```python
df.order_id = df.order_id.astype('int64')
```

```python
df.order_id.unique()
```

### customer_name

```python
df.customer_name = df.customer_name.str.strip()
```

```python
df.customer_name = df.customer_name.str.replace('Unknown','Unknown Last')
```

```python
df[['first_name', 'last_name']] = df['customer_name'].str.split(expand=True)
```

```python
col_name = df.columns.to_list()
col_name.remove('first_name')
col_name.remove('last_name')
col_name.insert(1,'first_name')
col_name.insert(2,'last_name')
df = df[col_name]
```

```python
df.drop('customer_name',axis = 1,inplace = True)
```
