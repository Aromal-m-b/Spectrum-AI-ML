```python
import pandas as pd
import numpy as np
```

```python
df['name'] = df['name'].str.capitalize()
```

```python
df['name'] = df['name'].str.strip()
```

```python
df['First Name'] = df['name'].apply(lambda x : x.split(' ')[0])
```

```python
df['Last Name'] = df['name'].apply(lambda x : x.split(' ')[1])
```

```python
df['Last Name'].isnull().sum()
```

```python
df.drop('name',axis = 1,inplace = True)
```

```python
cols = df.columns.to_list()
cols
```

```python
cols.remove('First Name')
cols.remove('Last Name')
cols.insert(1,'First Name')
cols.insert(2,'Last Name')
cols
```

```python
df = df[cols]
```

```python
df.head()
```

```python
df['email'].isnull().sum()
```

```python
df[df['email'].str[-10:]!='@gmail.com']
```

```python
df['email'] = df['email'].apply(
    lambda x: (
        x if pd.isna(x) or x.endswith('@gmail.com')
        else x.split('@')[0] + '@gmail.com'
    )
)
```

```python
df.head()
```

```python
df['email'] = df['email'].fillna(df['First Name'].str.lower()+'.'+df['Last Name'].str.lower()+'.'+'@gmail.com')
```

```python
df['age'] = df['age'].replace('unknown',np.nan) 
```

```python
df['age'] = df['age'].fillna(df['age'].mode()[0])
```

```python
df['age'] = df['age'].astype('int64')
```

```python
df.info()
```

```python
df['age'].unique()
```

```python
df.info()
```

```python
df['age'] = pd.to_numeric(df['age'], errors='coerce')
```

```python
df.loc[df['age'] > 120, 'age'] = df['age'].mode()[0]
```

```python
df.loc[df['age'] < 1,'age'] = df['age'].mode()[0]
```
