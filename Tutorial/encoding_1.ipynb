```python
from sklearn.preprocessing import OneHotEncoder

encoder = OneHotEncoder(
    sparse_output = False
)
type_encodes = encoder.fit_transform(cleaned[['type']])
type_features = encoder.get_feature_names_out(['type'])
types_df = pd.DataFrame(type_encodes,columns = type_features)
types_df
```

```python
test = pd.concat([cleaned.reset_index(drop = True),types_df.reset_index(drop = True)],axis = 1)
```

```python
cleaned = test.copy()
```

```python
rating_encodes = encoder.fit_transform(cleaned[['rating']])
rating_types = encoder.get_feature_names_out()
rating_df = pd.DataFrame(rating_encodes,columns = rating_types)
rating_df
```

```python
test = pd.concat([cleaned,rating_df],axis = 1)
cleaned = test.copy() 
```
