## Series

```python
data = [320,200,140,170]
series = pd.Series(data)
print(series)
```

### Labelling

```python
series = pd.Series(data,index=['x','y','z','o'])
print(series)
```

```python
print(series['z'])
```

### Seires using Dictionary

```python
data = {
    'x':320,
    'y':200,
    'z':140,
    'o':170
}
series = pd.Series(data)
print(series)
```
