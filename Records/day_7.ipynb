```python
import matplotlib.pyplot as plt
import numpy as np
import pandas as pd
```

```python
"""Draw a line to points (0,0) to (6,250)"""
x_points = np.array([0,6])
y_points = np.array([0,250])
plt.plot(x_points,y_points)
plt.show()
```

```python
"""Draw a line to points (1,3) to (8,10)"""
x_points = np.array([1,8])
y_points = np.array([3,10])
plt.plot(x_points,y_points)
plt.xlabel('x-axis')
plt.ylabel('y-axis')
plt.show()
```
